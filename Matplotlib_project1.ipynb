```python
import matplotlib.pyplot as plt
import pandas as pd
data=pd.read_csv("netflix_practice_dataset.csv")
print(data.head())
print(data.tail())
print(data.info())
data.dropna(subset=["score","rating","duration","genre","type","country","release_year","title","show_id"],inplace=True)
print(data.info())
print(data.shape)
print(data.describe())
```

```python
print(data.isnull().any())
print(data.isnull().sum())
print(data["country"].isnull().sum())
print(data["rating"].isnull().sum())
data.dropna(inplace=True)
cleaned_data=data.shape[0]
print(data.isnull().any())
print(cleaned_data)
```

```python
movies=data[data["type"]=="Movie"]
print(movies.head())
tv_shows=data[data["type"]=="TV Show"]
print(tv_shows.head())
```

```python
india=data[data["country"]=="India"]
print(india.head())
```

```python

after_2020=data[data["release_year"]>2020]
print(after_2020.head())
score_8_above=after_2020[after_2020["score"]>8]
print(score_8_above.head())
drama=data[data["genre"]=="Drama"]
print(drama.head())
all=data["type"].value_counts()
plt.pie(all.values,labels=all.index,autopct="%1.1f%%",colors=["red","blue"],explode=[0.1,0])
plt.title("Distribution of Movie and TV Show Types")
```

```python
plt.show()
plt.savefig("distribution_graph.png",dpi=300,bbox_inches="tight")

tvshows=data["type"].value_counts()
print(tvshows)

countries=data["country"].value_counts()
print(countries)

max_score=data["score"].max()
print(max_score)
mean_score=data["score"].mean()
print(mean_score)
```
