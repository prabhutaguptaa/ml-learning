```python
import matplotlib.pyplot as plt
x=["Mon", "Tue", "Wed", "Thu", "Fri"]
y=[12,13,35,7,11]
plt.bar(x,y,color="purple",edgecolor="black",label="Sales in this week")
plt.title("Backery Sales")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.legend(loc="upper left",fontsize=10)
plt.xlim("Mon", "Fri")
plt.ylim(0,40)
plt.xticks(["Mon", "Tue", "Wed", "Thu", "Fri"],["D1", "D2", "D3", "D4", "D5"])
plt.grid(color="black",linestyle=":",linewidth=0.5)
plt.show()
```

```python
import matplotlib.pyplot as plt
x=["Mon", "Tue", "Wed", "Thu", "Fri"]
y=[12,13,35,7,11]
plt.pie(y,labels=x,autopct="%1.1f%%",colors=["red","blue","green","yellow","orange"])
plt.title("Backery Sales")
plt.show()
```

```python
x=["Mon", "Tue", "Wed", "Thu", "Fri"]
y=[12,13,35,7,11]
plt.bar(x,y,color="blue",edgecolor="black",label="Sales in this week")
plt.title("Backery Sales")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.legend(loc="upper left",fontsize=10)
plt.xlim("Mon", "Fri")
plt.ylim(0,40)
plt.xticks(["Mon", "Tue", "Wed", "Thu", "Fri"],["D1", "D2", "D3", "D4", "D5"])
plt.grid(color="black",linestyle=":",linewidth=0.5)
plt.show()
```

```python
x=["Jan", "Feb", "Mar", "Apr", "May"]
y=[12,13,35,7,11]
fig,ax=plt.subplots(1,2,figsize=(12,5))
ax[0].plot(x,y)
ax[1].bar(x,y,color="red",edgecolor="black")
ax[0].set_title("Sales Data")
ax[1].set_title(" Data")
plt.tight_layout()
plt.suptitle("comparison of line and bar graph",fontsize=15)
plt.savefig("comparison_graph.png",dpi=300,bbox_inches="tight")
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[75,65,35,7,11]
plt.scatter(x,y,color="red",s=100,label="Students")
plt.scatter([3,4],[35,7],color="blue",s=100,label="Outsiders")
plt.title("Scattered Plot")
plt.xlabel("Hours Studied")
plt.ylabel("Test Scores")
plt.legend()
plt.show()
```
