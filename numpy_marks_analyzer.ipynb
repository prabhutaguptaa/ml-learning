Ek 2D array banao: 10 students ke 5 subjects ke marks (0 se 100). Random generate kar sakti ho, ya khud likh sakti ho. Shape (10, 5) hogi.

Tasks (har task ek alag cell mein, upar markdown mein likho ki kya kar rahi ho)

Array ki shape aur dtype print karo.
Har student ke total aur average marks nikalo.
Har subject ka average aur sabse zyada marks nikalo.
Topper kaun hai (student ka index)? Aur sabse kamzor subject kaun sa hai?
Wo saare students nikalo jinka average 60 se upar hai.
Jinke kisi subject mein 40 se kam marks hain, unke marks 40 se replace kar do (grace marks), original array safe rakhna.
Har subject ke marks mein 5% add karo, par 100 se upar nahi jana chahiye. Is task ke liye np.clip dhoondho.
Students ko average marks ke hisaab se descending order mein sort karo (argsort dekhna).

Bonus

```python
import numpy as np
# import python as py
rng=np.random.default_rng(42);
arr=rng.integers(low=0,high=100,size=(10,5));
print(arr);
```

```python
print(arr.shape)
```

```python
print(arr.dtype)
```

```python
total_marks=arr.sum(axis=1)
print(f"marks of each student is: {total_marks}")
```

```python
avg=arr.mean(axis=1)
print(f"average marks of each student is {avg}")
```

```python
print(f"average marks of each subject: {arr.mean(axis=0)}")
```

```python
print(f"maximum marks in each subject is: {arr.max(axis=0)}")
```

```python
print(f"index of tooper: {total_marks.argmax()}")
```

```python
print(f"index of lowest scorer: {total_marks.argmin()}")
```

```python
print(f"weak subject is:{arr.mean(axis=0).argmin()+1}")
```

```python
studenAbove60=total_marks[avg>60]
print(studenAbove60)
```

```python
arr1=arr.copy()
arr1[arr1<40]=40
print(arr1)
```

```python
desc_arr = np.sort(avg)[::-1]
print(desc_arr)
```

```python
student_order=np.argsort(avg)[::-1]
print(f"students higher to lower marks {student_order}")
```

```python
boosted = np.clip(arr * 1.05, 0, 100)
print(boosted)
```

```python
conditions = [avg >= 70, avg >= 55, avg >= 40]
grades = np.select(conditions, ['A', 'B', 'C'], default='F')
print(grades)
```
