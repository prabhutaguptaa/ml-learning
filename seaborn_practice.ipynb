```python
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
tips=sns.load_dataset("tips")
tips
```

```python
sns.scatterplot(data=tips,x='total_bill',y='tip')
```

```python
sns.relplot(data=tips,x='total_bill',y='tip',hue='sex',kind='scatter',style='time',size='size')
```

```python
gap=px.data.gapminder()
gap
```

```python
text_india=gap[gap["country"]=="India"]
text_india
```

```python
sns.lineplot(data=text_india,x="year",y="lifeExp")
```

```python
sns.relplot(data=text_india,x="year",y="lifeExp",kind="line")
```

```python
sns.relplot(data=text_india,x="year",y="lifeExp",size="gdpPercap")
```
